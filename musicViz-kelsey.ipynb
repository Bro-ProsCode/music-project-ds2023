## Load dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from ipywidgets import interact
import ipywidgets as widgets

spotify = pd.read_csv("spotify_tracks_with_release.csv")
spotify["original_release_year"] = spotify["original_release_year"].astype("Int64")

min_year = int(spotify["original_release_year"].min())
max_year = int(spotify["original_release_year"].max())

spotify.head()
```

```python
# words that clutter the visual
stopwords = {
    # version/edition
    "remastered", "remaster", "remix", "mix", "edit", "version", "live",
    "acoustic", "demo", "instrumental", "deluxe", "bonus", "mono", "stereo",
    "radio", "extended", "original", "reprise", "session", "single",
    # credits
    "feat", "ft", "featuring", "with", "vs",
    # function words
    "the", "and", "for", "you", "your", "from", "or", "el", "la", "los",
    "las", "del", "que", "les", "des", "une", "der", "die", "das",
    # misc
    "part", "pt", "vol", "track", "album",
}
```

```python
@interact(year=widgets.IntSlider(min=min_year, max=max_year, step=1,
                                 value=max_year, description="Year",
                                 continuous_update=False))
def plot_wordcloud(year):
    # Filter to the chosen year, then build the text
    subset = spotify.loc[spotify["original_release_year"] == year, "track_name"].dropna()

    if subset.empty:
        print(f"No tracks for {year}")
        return
        
    words = " ".join(subset.astype(str)).lower().split()
    textstr = " ".join(w for w in dict.fromkeys(words) if len(w) >= 3 and w not in stopwords)
    wordcloud = WordCloud(width=800, height=400, background_color="white").generate(textstr)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title(f"Track names from {year}")
    plt.show()
```

# How does a song’s acousticness relate to its release year?


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_tracks_with_release.csv")

df.head()
```

### What is the distribution of acoustiness among Spotify songs? 


```python

df["acousticness"].describe()
```

```python

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

df["acousticness"].plot(kind="hist", bins=40, ax=axes[0], edgecolor="white")
axes[0].set_title("Distribution of Acousticness")
axes[0].set_xlabel("Acousticness")

df["acousticness"].plot(kind="box", vert=False, ax=axes[1])
axes[1].set_title("Boxplot")

plt.tight_layout()
plt.show()

```

### How does the average acoustiness of songs change over the release years? 

```python

yearly = df.groupby("original_release_year")["acousticness"].mean()

plt.figure(figsize=(10, 5))
plt.plot(yearly.index, yearly.values)

plt.xlabel("Release Year")
plt.ylabel("Average Acousticness")
plt.title("Average Acousticness Over Time")

plt.show()

```
